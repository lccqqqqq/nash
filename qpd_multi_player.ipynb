## NE for 3 players on GHZ orbits

Looking at the amplitudes, the new family of NEs are not related to previous ones 

```python
state_0 = get_ghz_state(L=3)
H = get_default_H(num_players=3, option='H', dtype=np.float32)
results = []

for i in tqdm(range(10)):
    state = kick_with_u(state_0)
    state = to_canonical_form(state)
    result = find_nash_eq1(Psi=state, H=H, max_iter=1000, alpha=0.06, convergence_threshold=1e-6, expl_threshold=1e-3, use_tqdm=True, expl_check_interval=30, return_history=False)
    results.append(result)
```

```python
energies = np.array([result['energy'] for result in results])
print(energies)
```

```python
states = np.array([result['state'].reshape(-1) for result in results])
print(states)
```
