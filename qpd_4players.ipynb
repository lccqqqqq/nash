## Sample Random States

```python
L = 4 # number of players
def sample_random_states(L: int, num_samples: int, dtype: np.dtype = np.float32):
    """
    Sample random states from the unit sphere in the Hilbert space of the L-player game.
    """
    dim = 2**L
    if dtype == np.float32:
        states = np.random.randn(num_samples, dim)
    else:
        states = np.random.randn(num_samples, dim) + 1j * np.random.randn(num_samples, dim)
    states = states / np.linalg.norm(states, axis=-1, keepdims=True)
    return states

```
